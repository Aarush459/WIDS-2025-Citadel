```python
import numpy as np
import matplotlib.pyplot as plt
```

GBM Parameters


```python
S0 = 100      # initial stock price
mu = 0.10     # expected return (10%)
sigma = 0.20  # volatility (20%)
T = 1         # time horizon of 1 year
N = 252       # daily steps
dt = T / N
```

```python
paths = np.zeros((10, N + 1))
paths[:, 0] = S0

for i in range(10):
    for t in range(1, N + 1):
        Z = np.random.normal()   # Brownian noise
        paths[i, t] = paths[i, t-1] * np.exp((mu - 0.5*sigma**2)*dt + sigma*np.sqrt(dt)*Z)
```

```python
plt.figure(figsize=(12,6))
for i in range(10):
    plt.plot(paths[i], linewidth=1)
plt.title("GBM Simulated Stock Price Paths (10)")
plt.xlabel("Time Step")
plt.ylabel("Price")
plt.show()

```

```python
final_prices = paths[:, -1]

plt.figure(figsize=(8,4))
plt.hist(final_prices, bins=10, edgecolor="black")
plt.title("Histogram of Final Simulated Prices")
plt.xlabel("Final Price")
plt.ylabel("Frequency")
plt.show()
```

**Reflection**:



While running the GBM simulation, I noticed how strongly randomness affects the
future price of an asset. Even though all 10 simulated paths start from the exact
same initial price and follow the same drift and volatility parameters, the final
outcomes are very different. This happens because each step in the simulation
contains a random shock from Brownian motion, and these small random movements
keep accumulating over time. As a result, the price paths slowly drift apart and
form a wide distribution by the end of the simulation.

This widening of possible outcomes is essentially what we call risk in financial
markets. If the volatility value (sigma) is higher, the dispersion becomes even
larger, meaning the uncertainty about the final price increases. If volatility is
low, the paths stay closer together. So the simulation clearly shows the
relationship between randomness and risk: more randomness creates more possible
future scenarios, and therefore greater uncertainty.

Another interesting point is that randomness creates clustering effects. Some
paths show streaks of upward movement while others show streaks of decline, even
though none of this is intentional. It happens because random shocks often bunch
together. This reflects real markets, where long periods of calm or long periods
of turbulence naturally appear. Overall, the GBM experiment helped me understand
how unpredictable asset prices truly are and why modeling uncertainty is so
important in quant finance.
